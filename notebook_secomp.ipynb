## 4. Análise Exploratória e Visualização de Dados

Gráficos nos permitem enxergar padrões que tabelas numéricas escondem. Vamos construir visualizações para responder às nossas perguntas.

Vamos começar analisando a distribuição de massa corporal dos pinguins:

```python
sns.histplot(
    data=df_clean,
    x="body_mass_g",
    bins=20
)

plt.title("Distribuição da massa corporal")
plt.xlabel("Massa corporal (g)")
plt.ylabel("Número de pinguins")
plt.show()
```

Por espécie:

```python
sns.histplot(
    data=df_clean,
    x="body_mass_g",
    hue="species",
    bins=20
)

plt.title("Distribuição da massa corporal")
plt.xlabel("Massa corporal (g)")
plt.ylabel("Número de pinguins")
plt.show()
```

### ✏️ Passo 4.1: Contagem de Pinguins por Espécie

Crie um gráfico de barras para ver quantos pinguins de cada espécie foram analisados.
*Dica: Use `sns.countplot(data=df_clean, x='species')`.*

```python
# --- DIGITE SEU CÓDIGO AQUI ---
# Crie um gráfico de contagem para a coluna 'species'
sns.countplot(data=df_clean, x='species')
plt.title("Quantidade de Pinguins por Espécie")
plt.xlabel("Espécie")
plt.ylabel("Quantidade")
plt.show()
```

### ✏️ Passo 4.2: Comparando a Massa Corporal (Peso) com Boxplot

O **Boxplot** é ótimo para comparar distribuições de peso entre grupos e identificar valores discrepantes (*outliers*).

Crie um boxplot comparando o peso (`body_mass_g`) por espécie (`species`).
*Dica: Use `sns.boxplot(data=df_clean, x='species', y='body_mass_g')`.*

```python
# --- DIGITE SEU CÓDIGO AQUI ---
# Crie o boxplot de massa corporal por espécie
plt.title("Distribuição de Massa Corporal por Espécie")
plt.xlabel("Espécie")
plt.ylabel("Massa Corporal (g)")
plt.show()
```

###❓ O gráfico revelou algo que o dataframe não mostrou tão claramente?

Comparando a massa corporal, por espécie e sexo:


```python
sns.boxplot(
    data=df_clean,
    x="species",
    y="body_mass_g",
    hue="sex"
)

plt.title("Massa corporal por espécie e sexo")
plt.xlabel("Espécie")
plt.ylabel("Massa corporal (g)")
plt.show()
```

###❓As diferenças entre espécies permanecem quando consideramos o sexo?



### ✏️ Passo 4.3: Relação entre Comprimento e Profundidade do Bico

Vamos usar um **Gráfico de Dispersão (Scatter Plot)** para observar como o comprimento do bico se relaciona com a profundidade do bico, usando a cor (`hue`) para diferenciar as espécies.

*Dica: Use `sns.scatterplot(data=df_clean, x='bill_length_mm', y='bill_depth_mm', hue='species')`.*

```python
# --- DIGITE SEU CÓDIGO AQUI ---
# Crie o gráfico de dispersão do bico dos pinguins
plt.title("Comprimento vs. Profundidade do Bico por Espécie")
plt.xlabel("Comprimento do Bico (mm)")
plt.ylabel("Profundidade do Bico (mm)")
plt.show()
```

### Tornando a análise interativa com o Plotly

Utilizando a biblioteca Plotly, criamos gráficos interativos que exibem mais detalhes ao passar o mouse sobre os dados.


```python
import plotly.express as px
```

```python
fig = px.scatter(
    df_clean,
    x="flipper_length_mm",
    y="body_mass_g",
    color="species",
    symbol="sex",
    hover_data=[
        "island",
        "bill_length_mm",
        "bill_depth_mm"
    ],
    title="Massa corporal × comprimento da nadadeira"
)

fig.show()
```
